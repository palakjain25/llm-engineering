```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.tools import create_retriever_tool
from langchain.agents import  create_agent
from langchain_openrouter import ChatOpenRouter
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
texts = [
    'Apple makes very good computers',
    'I believe apple is innovative!',
    'I love apples',
    'I am a fan of macbooks',
    'I enjoy oranges',
    'I like lenovo thinkpads',
    'i think pears taste very good',
    'I despise guava'
]
```

```python
vector_stores = FAISS.from_texts(texts, embedding=embeddings)
```

```python
vector_stores.similarity_search('Applie is my favourite fruit', k=7)
```

```python
vector_stores.similarity_search('Linus is good OS', k=7)
```

```python
vector_stores.similarity_search('what fruit does the person not like?', k=7)
```

```python
retriever = vector_stores.as_retriever(search_kwargs={'k':3})
retriever_tool = create_retriever_tool(retriever , name = 'kb_search', description = 'Search the small product/fruit database for information.')
```

```python
model = ChatOpenRouter(
    model="openrouter/free",
    temperature=0.2,
)
```

```python
agent = create_agent(
    model = model,
    tools = [retriever_tool],
    system_prompt = (
        "You are a helpful assistant. For questions about Macs, apples, or laptops"
        "first call the kb_Search tool to retrieve context, then answer succinctly. Maybe you have to use it multiple times before answering."
    )
)
```

```python
result = agent.invoke({
    "messages" : [
        {
            "role" : "user",
            "content" : "what three fruits does the person like and what three fruits does the person dislike?"
        }
    ]
})
```
