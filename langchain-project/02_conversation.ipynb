```python
import requests
from dotenv import load_dotenv

# from langchain.chat_models import init_chat_model
from langchain_openrouter import ChatOpenRouter
from langchain.messages import HumanMessage, AIMessage, SystemMessage

load_dotenv()
```

```python
model = ChatOpenRouter(
    model="openrouter/free",
    temperature=0.2,
)
```

```python
conversation = [
    SystemMessage('You are a helpful assistant for questions regarding programming'),
    HumanMessage('What is python?'),
    AIMessage('Python is an interprested programming language.'),
    HumanMessage('When was it released?')
]
```

```python
# response = model.invoke('Hello, what is python?')
response = model.invoke(conversation)
```

```python
print(response)
```
