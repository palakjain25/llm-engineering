```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import  create_agent
from langchain_openrouter import ChatOpenRouter

from langchain.agents.middleware import SummarizationMiddleware
```

```python
model = ChatOpenRouter(
    model="openrouter/free",
    temperature=0.2,
)
```

```python
agent =create_agent(
    model=model,
    middleware=[
        SummarizationMiddleware(
            model = model,
            max_tokens_before_summary = 4000,
            messages_to_keep = 20,
            summary_prompt = 'Summarize thee most important keypoint from this conversation.'
        )
    ]
)
```

```python
response = agent.invoke({
    'messages':[
        {'role' : 'user', 'content' : 'Retrieval-Augmented Generation (RAG) is a technique that enhances the responses of large language models (LLMs) by retrieving relevant information from external data sources before generating an answer. Instead of relying solely on the knowledge learned during training, a RAG system searches a knowledge base, retrieves relevant documents or text chunks, and provides them as context to the LLM. This helps generate more accurate, relevant, and up-to-date responses while reducing hallucinations. RAG is commonly used in chatbots, document question-answering systems, and AI agents that need to summarize conversations or answer questions based on previous interactions.'}
    ]
})
```

```python
print(response)
```
